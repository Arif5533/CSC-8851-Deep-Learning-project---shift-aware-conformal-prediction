```python
import os
import csv
import numpy as np
import matplotlib.pyplot as plt

try:
    from tqdm import tqdm
except ImportError:
    def tqdm(x, **kwargs):
        return x

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, random_split
from torchvision import datasets, transforms
from torchvision.models import resnet18

# ============================================================
# Paths
# ============================================================
PROJECT_ROOT = os.getcwd()
DATA_ROOT = os.path.join(PROJECT_ROOT, "shift_aware_conformal_data")
OUTPUT_ROOT = os.path.join(PROJECT_ROOT, "shift_aware_conformal_outputs")
FIG_DIR = os.path.join(OUTPUT_ROOT, "step7_edl_extension_figures")

os.makedirs(FIG_DIR, exist_ok=True)

BASELINE_MODEL_PATH = os.path.join(OUTPUT_ROOT, "resnet18_cifar10_best_debug.pt")
BASELINE_CAL_PATH = os.path.join(OUTPUT_ROOT, "cal_outputs_full.npz")
BASELINE_TEST_PATH = os.path.join(OUTPUT_ROOT, "test_outputs_full.npz")

EDL_MODEL_PATH = os.path.join(OUTPUT_ROOT, "resnet18_cifar10_edl_best.pt")
EDL_CAL_PATH = os.path.join(OUTPUT_ROOT, "edl_cal_outputs.npz")
EDL_TEST_PATH = os.path.join(OUTPUT_ROOT, "edl_test_outputs.npz")

GATE_SUMMARY_CSV = os.path.join(OUTPUT_ROOT, "step7_edl_gate_summary.csv")
GATE_SHIFT_CSV = os.path.join(OUTPUT_ROOT, "step7_edl_gate_cifar10c.csv")
EDL_RAPS_CSV = os.path.join(OUTPUT_ROOT, "step7_edl_raps_clean.csv")
RESULTS_NPZ = os.path.join(OUTPUT_ROOT, "step7_edl_extension_results.npz")

CIFAR10C_ROOT = os.path.join(DATA_ROOT, "CIFAR-10-C")

print("PROJECT_ROOT       :", PROJECT_ROOT)
print("DATA_ROOT          :", DATA_ROOT)
print("OUTPUT_ROOT        :", OUTPUT_ROOT)
print("FIG_DIR            :", FIG_DIR)
print("BASELINE_MODEL     :", BASELINE_MODEL_PATH)
print("BASELINE_CAL       :", BASELINE_CAL_PATH)
print("BASELINE_TEST      :", BASELINE_TEST_PATH)
print("EDL_MODEL          :", EDL_MODEL_PATH)
print("CIFAR10C_ROOT      :", CIFAR10C_ROOT)

# ============================================================
# Checks
# ============================================================
for path in [BASELINE_MODEL_PATH, BASELINE_CAL_PATH, BASELINE_TEST_PATH]:
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing required baseline file: {path}")

if not os.path.exists(CIFAR10C_ROOT):
    raise FileNotFoundError(f"Missing CIFAR-10-C folder: {CIFAR10C_ROOT}")

# ============================================================
# Device / settings
# ============================================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device             :", device)

SEED = 42
BATCH_SIZE = 128
NUM_WORKERS = 0
NUM_CLASSES = 10

# EDL training
EPOCHS = 20
LR = 1e-3
WEIGHT_DECAY = 1e-4
ANNEALING_STEP = 10

# Conformal / gate settings
ALPHAS = [0.10, 0.05]
DELTAS = [0.05, 0.10]

# Keep same RAPS settings as your pipeline
LAMBDA_REG = 0.01
K_REG = 2
RANDOMIZED = True

CORRUPTIONS = [
    "gaussian_noise",
    "motion_blur",
    "brightness",
    "jpeg_compression",
]
SEVERITIES = [1, 2, 3, 4, 5]

torch.manual_seed(SEED)
np.random.seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# ============================================================
# Models
# ============================================================
class ResNet18Softmax(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.backbone = resnet18(weights=None)
        self.backbone.conv1 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)
        self.backbone.maxpool = nn.Identity()
        in_features = self.backbone.fc.in_features
        self.backbone.fc = nn.Linear(in_features, num_classes)

    def forward(self, x):
        return self.backbone(x)

class ResNet18EDL(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.backbone = resnet18(weights=None)
        self.backbone.conv1 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)
        self.backbone.maxpool = nn.Identity()
        in_features = self.backbone.fc.in_features
        self.backbone.fc = nn.Linear(in_features, num_classes)
        self.softplus = nn.Softplus()

    def forward(self, x):
        logits = self.backbone(x)
        evidence = self.softplus(logits)
        alpha = evidence + 1.0
        return alpha

# ============================================================
# CIFAR-10-C dataset
# ============================================================
class CIFAR10CDataset(Dataset):
    def __init__(self, cifar10c_root, corruption_name, severity, transform=None):
        self.transform = transform

        corruption_path = os.path.join(cifar10c_root, f"{corruption_name}.npy")
        labels_path = os.path.join(cifar10c_root, "labels.npy")

        if not os.path.exists(corruption_path):
            raise FileNotFoundError(f"Missing corruption file: {corruption_path}")
        if not os.path.exists(labels_path):
            raise FileNotFoundError(f"Missing labels file: {labels_path}")

        all_images = np.load(corruption_path)
        all_labels = np.load(labels_path)

        start = (severity - 1) * 10000
        end = severity * 10000

        self.images = all_images[start:end]
        self.labels = all_labels[start:end]

        assert len(self.images) == 10000
        assert len(self.labels) == 10000

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        img = self.images[idx]
        label = int(self.labels[idx])

        if self.transform is not None:
            img = self.transform(img)

        return img, label

# ============================================================
# Transforms / datasets
# ============================================================
transform_train = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=(0.4914, 0.4822, 0.4465),
        std=(0.2470, 0.2435, 0.2616),
    ),
])

transform_eval = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(
        mean=(0.4914, 0.4822, 0.4465),
        std=(0.2470, 0.2435, 0.2616),
    ),
])

transform_cifar10c = transforms.Compose([
    transforms.ToPILImage(),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=(0.4914, 0.4822, 0.4465),
        std=(0.2470, 0.2435, 0.2616),
    ),
])

cifar10_full = datasets.CIFAR10(
    root=DATA_ROOT, train=True, download=True, transform=transform_train
)
cifar10_full_eval = datasets.CIFAR10(
    root=DATA_ROOT, train=True, download=True, transform=transform_eval
)
cifar10_test = datasets.CIFAR10(
    root=DATA_ROOT, train=False, download=True, transform=transform_eval
)
svhn_test = datasets.SVHN(
    root=DATA_ROOT, split="test", download=True, transform=transform_eval
)

g = torch.Generator().manual_seed(SEED)
train_size = 45000
cal_size = 5000
train_subset, cal_subset = random_split(cifar10_full, [train_size, cal_size], generator=g)
_, cal_subset_eval = random_split(cifar10_full_eval, [train_size, cal_size], generator=torch.Generator().manual_seed(SEED))

train_loader = DataLoader(train_subset, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS)
cal_loader = DataLoader(cal_subset_eval, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS)
test_loader = DataLoader(cifar10_test, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS)
svhn_loader = DataLoader(svhn_test, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS)

print("Train size         :", len(train_subset))
print("Cal size           :", len(cal_subset_eval))
print("Test size          :", len(cifar10_test))
print("SVHN size          :", len(svhn_test))

# ============================================================
# EDL loss
# ============================================================
def one_hot(labels, num_classes):
    return F.one_hot(labels, num_classes=num_classes).float()

def kl_dirichlet(alpha, num_classes):
    beta = torch.ones((1, num_classes), device=alpha.device)
    S_alpha = torch.sum(alpha, dim=1, keepdim=True)
    S_beta = torch.sum(beta, dim=1, keepdim=True)

    lnB_alpha = torch.lgamma(S_alpha) - torch.sum(torch.lgamma(alpha), dim=1, keepdim=True)
    lnB_beta = torch.sum(torch.lgamma(beta), dim=1, keepdim=True) - torch.lgamma(S_beta)

    dg0 = torch.digamma(S_alpha)
    dg1 = torch.digamma(alpha)

    kl = torch.sum((alpha - beta) * (dg1 - dg0), dim=1, keepdim=True) + lnB_alpha + lnB_beta
    return kl

def edl_mse_loss(alpha, target, epoch, num_classes, annealing_step):
    S = torch.sum(alpha, dim=1, keepdim=True)
    mean = alpha / S

    err = torch.sum((target - mean) ** 2, dim=1, keepdim=True)
    var = torch.sum(alpha * (S - alpha) / (S * S * (S + 1.0)), dim=1, keepdim=True)

    annealing_coef = min(1.0, epoch / annealing_step)
    alpha_tilde = (alpha - 1.0) * (1.0 - target) + 1.0
    kl = annealing_coef * kl_dirichlet(alpha_tilde, num_classes)

    return torch.mean(err + var + kl)

# ============================================================
# Train / eval helpers
# ============================================================
def edl_accuracy(alpha, labels):
    probs = alpha / alpha.sum(dim=1, keepdim=True)
    preds = probs.argmax(dim=1)
    return (preds == labels).float().mean().item()

def evaluate_edl(model, loader, device, epoch=1):
    model.eval()
    total_loss = 0.0
    total_acc = 0.0
    total_n = 0

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)

            alpha = model(images)
            target = one_hot(labels, NUM_CLASSES)
            loss = edl_mse_loss(alpha, target, epoch, NUM_CLASSES, ANNEALING_STEP)

            bs = labels.size(0)
            total_loss += loss.item() * bs
            total_acc += edl_accuracy(alpha, labels) * bs
            total_n += bs

    return total_loss / total_n, total_acc / total_n

def train_edl_model():
    model = ResNet18EDL(num_classes=NUM_CLASSES).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)

    best_val_acc = -1.0
    history = []

    for epoch in range(1, EPOCHS + 1):
        model.train()
        running_loss = 0.0
        running_acc = 0.0
        running_n = 0

        for images, labels in tqdm(train_loader, desc=f"EDL train epoch {epoch}", leave=False):
            images = images.to(device)
            labels = labels.to(device)

            optimizer.zero_grad()
            alpha = model(images)
            target = one_hot(labels, NUM_CLASSES)
            loss = edl_mse_loss(alpha, target, epoch, NUM_CLASSES, ANNEALING_STEP)
            loss.backward()
            optimizer.step()

            bs = labels.size(0)
            running_loss += loss.item() * bs
            running_acc += edl_accuracy(alpha, labels) * bs
            running_n += bs

        train_loss = running_loss / running_n
        train_acc = running_acc / running_n
        val_loss, val_acc = evaluate_edl(model, cal_loader, device, epoch=epoch)

        history.append({
            "epoch": epoch,
            "train_loss": train_loss,
            "train_acc": train_acc,
            "val_loss": val_loss,
            "val_acc": val_acc,
        })

        print(
            f"Epoch {epoch:02d} | "
            f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} | "
            f"val_loss={val_loss:.4f} val_acc={val_acc:.4f}"
        )

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), EDL_MODEL_PATH)
            print(f"  Saved best EDL model to {EDL_MODEL_PATH}")

    return history

# ============================================================
# Train or load EDL model
# ============================================================
if not os.path.exists(EDL_MODEL_PATH):
    print("\nTraining EDL model from scratch...")
    train_history = train_edl_model()
else:
    print("\nFound existing EDL model, skipping training.")
    train_history = None

edl_model = ResNet18EDL(num_classes=NUM_CLASSES).to(device)
edl_model.load_state_dict(torch.load(EDL_MODEL_PATH, map_location=device))
edl_model.eval()
print("Loaded EDL model successfully.")

# ============================================================
# Collect EDL outputs
# ============================================================
def collect_edl_outputs(model, loader, device, desc="Collect EDL"):
    model.eval()
    all_mean_probs = []
    all_alpha0 = []
    all_labels = []

    with torch.no_grad():
        for images, labels in tqdm(loader, desc=desc, leave=False):
            images = images.to(device)
            alpha = model(images)
            alpha0 = alpha.sum(dim=1)
            mean_probs = alpha / alpha.sum(dim=1, keepdim=True)

            all_mean_probs.append(mean_probs.cpu().numpy())
            all_alpha0.append(alpha0.cpu().numpy())
            all_labels.append(labels.numpy())

    return (
        np.concatenate(all_mean_probs, axis=0),
        np.concatenate(all_alpha0, axis=0),
        np.concatenate(all_labels, axis=0),
    )

# Save EDL cal/test outputs
if (not os.path.exists(EDL_CAL_PATH)) or (not os.path.exists(EDL_TEST_PATH)):
    edl_cal_probs, edl_cal_alpha0, edl_cal_labels = collect_edl_outputs(edl_model, cal_loader, device, desc="EDL cal")
    edl_test_probs, edl_test_alpha0, edl_test_labels = collect_edl_outputs(edl_model, test_loader, device, desc="EDL test")

    np.savez_compressed(
        EDL_CAL_PATH,
        probs=edl_cal_probs,
        alpha0=edl_cal_alpha0,
        labels=edl_cal_labels,
    )
    np.savez_compressed(
        EDL_TEST_PATH,
        probs=edl_test_probs,
        alpha0=edl_test_alpha0,
        labels=edl_test_labels,
    )
    print("Saved EDL outputs:")
    print(" -", EDL_CAL_PATH)
    print(" -", EDL_TEST_PATH)

edl_cal = np.load(EDL_CAL_PATH)
edl_test = np.load(EDL_TEST_PATH)

edl_cal_probs = edl_cal["probs"]
edl_cal_alpha0 = edl_cal["alpha0"]
edl_cal_labels = edl_cal["labels"]

edl_test_probs = edl_test["probs"]
edl_test_alpha0 = edl_test["alpha0"]
edl_test_labels = edl_test["labels"]

print("\nLoaded EDL arrays:")
print("edl_cal_probs   :", edl_cal_probs.shape)
print("edl_cal_alpha0  :", edl_cal_alpha0.shape)
print("edl_test_probs  :", edl_test_probs.shape)
print("edl_test_alpha0 :", edl_test_alpha0.shape)

# ============================================================
# RAPS helpers (same as your pipeline, but on EDL mean probs)
# ============================================================
def conformal_quantile(scores, alpha):
    n = len(scores)
    k = int(np.ceil((n + 1) * (1 - alpha))) - 1
    k = min(max(k, 0), n - 1)
    return np.sort(scores)[k]

def compute_raps_score(prob_vec, true_label, lambda_reg=0.0, k_reg=0, randomized=True, rng=None):
    if rng is None:
        rng = np.random.default_rng()

    sorted_idx = np.argsort(-prob_vec)
    sorted_probs = prob_vec[sorted_idx]
    L = int(np.where(sorted_idx == true_label)[0][0]) + 1

    score = np.sum(sorted_probs[:L]) + lambda_reg * max(L - k_reg, 0)
    if randomized:
        u = rng.uniform()
        score -= u * sorted_probs[L - 1]
    return float(score)

def compute_all_raps_scores(probs, labels, lambda_reg=0.0, k_reg=0, randomized=True, seed=0):
    rng = np.random.default_rng(seed)
    return np.array([
        compute_raps_score(
            probs[i],
            labels[i],
            lambda_reg=lambda_reg,
            k_reg=k_reg,
            randomized=randomized,
            rng=rng
        )
        for i in range(len(labels))
    ], dtype=np.float64)

def build_raps_set(prob_vec, qhat, lambda_reg=0.0, k_reg=0, randomized=True, rng=None):
    if rng is None:
        rng = np.random.default_rng()

    sorted_idx = np.argsort(-prob_vec)
    sorted_probs = prob_vec[sorted_idx]
    K = len(sorted_probs)

    penalized_cum = np.array([
        np.sum(sorted_probs[:j]) + lambda_reg * max(j - k_reg, 0)
        for j in range(1, K + 1)
    ], dtype=np.float64)

    count_leq = int(np.sum(penalized_cum <= qhat))
    if count_leq >= K:
        return [int(x) for x in sorted_idx]

    L = count_leq + 1

    if randomized:
        boundary_value = penalized_cum[L - 1]
        increment = sorted_probs[L - 1] + (lambda_reg if L > k_reg else 0.0)
        drop_prob = (boundary_value - qhat) / max(increment, 1e-12)
        drop_prob = float(np.clip(drop_prob, 0.0, 1.0))

        u = rng.uniform()
        if u <= drop_prob:
            L -= 1

    return [int(x) for x in sorted_idx[:L]]

def build_all_prediction_sets(probs, qhat, lambda_reg=0.0, k_reg=0, randomized=True, seed=0):
    rng = np.random.default_rng(seed)
    return [
        build_raps_set(
            prob_vec,
            qhat=qhat,
            lambda_reg=lambda_reg,
            k_reg=k_reg,
            randomized=randomized,
            rng=rng
        )
        for prob_vec in probs
    ]

def empirical_coverage(pred_sets, labels):
    return float(np.mean([int(y) in s for s, y in zip(pred_sets, labels)]))

def average_set_size(pred_sets):
    return float(np.mean([len(s) for s in pred_sets]))

# ============================================================
# Clean EDL-RAPS
# ============================================================
edl_raps_rows = []
edl_cal_scores = compute_all_raps_scores(
    edl_cal_probs,
    edl_cal_labels,
    lambda_reg=LAMBDA_REG,
    k_reg=K_REG,
    randomized=RANDOMIZED,
    seed=SEED,
)

print("\n" + "=" * 80)
print("CLEAN EDL-RAPS RESULTS")
print("=" * 80)

for alpha in ALPHAS:
    qhat = conformal_quantile(edl_cal_scores, alpha)
    pred_sets = build_all_prediction_sets(
        edl_test_probs,
        qhat=qhat,
        lambda_reg=LAMBDA_REG,
        k_reg=K_REG,
        randomized=RANDOMIZED,
        seed=SEED + 100,
    )
    cov = empirical_coverage(pred_sets, edl_test_labels)
    size = average_set_size(pred_sets)

    row = {
        "method": "EDL-RAPS",
        "alpha": alpha,
        "target_coverage": 1 - alpha,
        "qhat": qhat,
        "empirical_coverage": cov,
        "coverage_gap": cov - (1 - alpha),
        "avg_set_size": size,
    }
    edl_raps_rows.append(row)

    print(
        f"alpha={alpha:.2f} | qhat={qhat:.6f} | "
        f"coverage={cov:.4f} | gap={cov - (1-alpha):+.4f} | size={size:.4f}"
    )

with open(EDL_RAPS_CSV, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=list(edl_raps_rows[0].keys()))
    writer.writeheader()
    writer.writerows(edl_raps_rows)

# ============================================================
# Baseline energy scores
# ============================================================
baseline_cal = np.load(BASELINE_CAL_PATH)
baseline_test = np.load(BASELINE_TEST_PATH)

baseline_cal_logits = baseline_cal["logits"]
baseline_test_logits = baseline_test["logits"]

def energy_score_from_logits(logits, T=1.0):
    logits_t = torch.tensor(logits, dtype=torch.float32) / T
    energy = -T * torch.logsumexp(logits_t, dim=1)
    return energy.numpy()

energy_cal = energy_score_from_logits(baseline_cal_logits, T=1.0)
energy_test = energy_score_from_logits(baseline_test_logits, T=1.0)

# EDL gate score: larger = more OOD-like
edl_gate_cal = -edl_cal_alpha0
edl_gate_test = -edl_test_alpha0

# ============================================================
# Need baseline model for SVHN / CIFAR-10-C energy
# ============================================================
baseline_model = ResNet18Softmax(num_classes=NUM_CLASSES).to(device)
baseline_model.load_state_dict(torch.load(BASELINE_MODEL_PATH, map_location=device))
baseline_model.eval()

def collect_softmax_logits(model, loader, device, desc="Collect logits"):
    model.eval()
    all_logits = []
    all_labels = []

    with torch.no_grad():
        for images, labels in tqdm(loader, desc=desc, leave=False):
            images = images.to(device)
            logits = model(images)
            all_logits.append(logits.cpu().numpy())
            all_labels.append(labels.numpy())

    return np.concatenate(all_logits, axis=0), np.concatenate(all_labels, axis=0)

# SVHN scores
svhn_logits, svhn_labels = collect_softmax_logits(baseline_model, svhn_loader, device, desc="Baseline SVHN logits")
energy_svhn = energy_score_from_logits(svhn_logits, T=1.0)

edl_svhn_probs, edl_svhn_alpha0, edl_svhn_labels = collect_edl_outputs(edl_model, svhn_loader, device, desc="EDL SVHN")
edl_gate_svhn = -edl_svhn_alpha0

# ============================================================
# Combined gate
# ============================================================
def zscore_with_cal(scores, cal_scores):
    mu = np.mean(cal_scores)
    std = np.std(cal_scores) + 1e-12
    return (scores - mu) / std

combined_cal = zscore_with_cal(energy_cal, energy_cal) + zscore_with_cal(edl_gate_cal, edl_gate_cal)
combined_test = zscore_with_cal(energy_test, energy_cal) + zscore_with_cal(edl_gate_test, edl_gate_cal)
combined_svhn = zscore_with_cal(energy_svhn, energy_cal) + zscore_with_cal(edl_gate_svhn, edl_gate_cal)

# ============================================================
# Conformal gate thresholding
# ============================================================
def conformal_upper_quantile(scores, delta):
    n = len(scores)
    k = int(np.ceil((n + 1) * (1 - delta))) - 1
    k = min(max(k, 0), n - 1)
    return np.sort(scores)[k]

def trigger_rate(scores, threshold):
    return float(np.mean(scores >= threshold))

gate_summary_rows = []

print("\n" + "=" * 80)
print("GATE SUMMARY: CLEAN ID vs SVHN")
print("=" * 80)

for delta in DELTAS:
    thr_energy = conformal_upper_quantile(energy_cal, delta)
    thr_edl = conformal_upper_quantile(edl_gate_cal, delta)
    thr_combined = conformal_upper_quantile(combined_cal, delta)

    rows = [
        {
            "gate": "Energy",
            "delta": delta,
            "dataset": "clean_test_id",
            "threshold": thr_energy,
            "trigger_rate": trigger_rate(energy_test, thr_energy),
        },
        {
            "gate": "Energy",
            "delta": delta,
            "dataset": "svhn_ood",
            "threshold": thr_energy,
            "trigger_rate": trigger_rate(energy_svhn, thr_energy),
        },
        {
            "gate": "EDL",
            "delta": delta,
            "dataset": "clean_test_id",
            "threshold": thr_edl,
            "trigger_rate": trigger_rate(edl_gate_test, thr_edl),
        },
        {
            "gate": "EDL",
            "delta": delta,
            "dataset": "svhn_ood",
            "threshold": thr_edl,
            "trigger_rate": trigger_rate(edl_gate_svhn, thr_edl),
        },
        {
            "gate": "Combined",
            "delta": delta,
            "dataset": "clean_test_id",
            "threshold": thr_combined,
            "trigger_rate": trigger_rate(combined_test, thr_combined),
        },
        {
            "gate": "Combined",
            "delta": delta,
            "dataset": "svhn_ood",
            "threshold": thr_combined,
            "trigger_rate": trigger_rate(combined_svhn, thr_combined),
        },
    ]

    gate_summary_rows.extend(rows)

for row in gate_summary_rows:
    print(
        f"{row['gate']:8s} | delta={row['delta']:.2f} | "
        f"{row['dataset']:12s} | trigger={row['trigger_rate']:.4f}"
    )

# ============================================================
# CIFAR-10-C gate evaluation
# ============================================================
gate_shift_rows = []

for corruption in CORRUPTIONS:
    print("\n" + "=" * 80)
    print("Corruption:", corruption)

    for severity in SEVERITIES:
        dataset = CIFAR10CDataset(
            cifar10c_root=CIFAR10C_ROOT,
            corruption_name=corruption,
            severity=severity,
            transform=transform_cifar10c,
        )

        loader = DataLoader(
            dataset,
            batch_size=BATCH_SIZE,
            shuffle=False,
            num_workers=NUM_WORKERS,
        )

        logits_shift, labels_shift = collect_softmax_logits(
            baseline_model, loader, device, desc=f"Energy {corruption} sev{severity}"
        )
        energy_shift = energy_score_from_logits(logits_shift, T=1.0)

        _, edl_alpha0_shift, _ = collect_edl_outputs(
            edl_model, loader, device, desc=f"EDL {corruption} sev{severity}"
        )
        edl_gate_shift = -edl_alpha0_shift

        combined_shift = (
            zscore_with_cal(energy_shift, energy_cal)
            + zscore_with_cal(edl_gate_shift, edl_gate_cal)
        )

        for delta in DELTAS:
            thr_energy = conformal_upper_quantile(energy_cal, delta)
            thr_edl = conformal_upper_quantile(edl_gate_cal, delta)
            thr_combined = conformal_upper_quantile(combined_cal, delta)

            gate_shift_rows.append({
                "corruption": corruption,
                "severity": severity,
                "delta": delta,
                "gate": "Energy",
                "trigger_rate": trigger_rate(energy_shift, thr_energy),
            })
            gate_shift_rows.append({
                "corruption": corruption,
                "severity": severity,
                "delta": delta,
                "gate": "EDL",
                "trigger_rate": trigger_rate(edl_gate_shift, thr_edl),
            })
            gate_shift_rows.append({
                "corruption": corruption,
                "severity": severity,
                "delta": delta,
                "gate": "Combined",
                "trigger_rate": trigger_rate(combined_shift, thr_combined),
            })

        print(f"severity={severity} finished")

# ============================================================
# Save CSV / NPZ
# ============================================================
with open(GATE_SUMMARY_CSV, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=list(gate_summary_rows[0].keys()))
    writer.writeheader()
    writer.writerows(gate_summary_rows)

with open(GATE_SHIFT_CSV, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=list(gate_shift_rows[0].keys()))
    writer.writeheader()
    writer.writerows(gate_shift_rows)

np.savez_compressed(
    RESULTS_NPZ,
    edl_raps_rows=np.array(edl_raps_rows, dtype=object),
    gate_summary_rows=np.array(gate_summary_rows, dtype=object),
    gate_shift_rows=np.array(gate_shift_rows, dtype=object),
)

print("\nSaved:")
print(" -", EDL_RAPS_CSV)
print(" -", GATE_SUMMARY_CSV)
print(" -", GATE_SHIFT_CSV)
print(" -", RESULTS_NPZ)

# ============================================================
# Figures
# ============================================================
# Figure 1: Clean ID vs SVHN trigger bar chart
for delta in DELTAS:
    gates = ["Energy", "EDL", "Combined"]
    clean_rates = []
    svhn_rates = []

    for gate in gates:
        clean_rate = next(
            r["trigger_rate"] for r in gate_summary_rows
            if r["gate"] == gate and abs(r["delta"] - delta) < 1e-12 and r["dataset"] == "clean_test_id"
        )
        svhn_rate = next(
            r["trigger_rate"] for r in gate_summary_rows
            if r["gate"] == gate and abs(r["delta"] - delta) < 1e-12 and r["dataset"] == "svhn_ood"
        )
        clean_rates.append(clean_rate)
        svhn_rates.append(svhn_rate)

    x = np.arange(len(gates))
    width = 0.35

    plt.figure(figsize=(8, 5))
    plt.bar(x - width/2, clean_rates, width, label="Clean test ID")
    plt.bar(x + width/2, svhn_rates, width, label="SVHN OOD")
    plt.xticks(x, gates)
    plt.ylabel("Trigger rate")
    plt.title(f"Gate comparison on clean ID vs SVHN (delta={delta:.2f})")
    plt.ylim(0, 1.0)
    plt.grid(True, axis="y", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(
        os.path.join(FIG_DIR, f"bar_gate_compare_delta_{str(delta).replace('.', 'p')}.png"),
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()

# Figure 2: Trigger vs severity for each gate
for delta in DELTAS:
    for gate in ["Energy", "EDL", "Combined"]:
        plt.figure(figsize=(8, 5))

        for corruption in CORRUPTIONS:
            rows = [
                r for r in gate_shift_rows
                if r["gate"] == gate and abs(r["delta"] - delta) < 1e-12 and r["corruption"] == corruption
            ]
            rows = sorted(rows, key=lambda x: x["severity"])
            xs = [r["severity"] for r in rows]
            ys = [r["trigger_rate"] for r in rows]
            plt.plot(xs, ys, marker="o", label=corruption)

        plt.xlabel("Severity")
        plt.ylabel("Trigger rate")
        plt.title(f"{gate} gate trigger vs severity (delta={delta:.2f})")
        plt.xticks(SEVERITIES)
        plt.ylim(0, 1.0)
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.savefig(
            os.path.join(FIG_DIR, f"{gate.lower()}_trigger_vs_severity_delta_{str(delta).replace('.', 'p')}.png"),
            dpi=300,
            bbox_inches="tight"
        )
        plt.show()

# Figure 3: Histograms for gate scores on clean/SVHN
plt.figure(figsize=(9, 6))
plt.hist(energy_test, bins=60, alpha=0.4, density=True, label="Energy clean ID")
plt.hist(energy_svhn, bins=60, alpha=0.4, density=True, label="Energy SVHN")
plt.xlabel("Energy score")
plt.ylabel("Density")
plt.title("Energy score distributions")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "hist_energy_clean_vs_svhn.png"), dpi=300, bbox_inches="tight")
plt.show()

plt.figure(figsize=(9, 6))
plt.hist(edl_gate_test, bins=60, alpha=0.4, density=True, label="EDL clean ID")
plt.hist(edl_gate_svhn, bins=60, alpha=0.4, density=True, label="EDL SVHN")
plt.xlabel("EDL gate score (-alpha0)")
plt.ylabel("Density")
plt.title("EDL gate score distributions")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "hist_edl_clean_vs_svhn.png"), dpi=300, bbox_inches="tight")
plt.show()

plt.figure(figsize=(9, 6))
plt.hist(combined_test, bins=60, alpha=0.4, density=True, label="Combined clean ID")
plt.hist(combined_svhn, bins=60, alpha=0.4, density=True, label="Combined SVHN")
plt.xlabel("Combined normalized score")
plt.ylabel("Density")
plt.title("Combined gate score distributions")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, "hist_combined_clean_vs_svhn.png"), dpi=300, bbox_inches="tight")
plt.show()

print("\nSaved figures in:", FIG_DIR)
print("Step 7 EDL extension finished.")
```
