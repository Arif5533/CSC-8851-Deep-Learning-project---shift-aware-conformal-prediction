```python
import os
import csv
import numpy as np
import matplotlib.pyplot as plt

try:
    from tqdm import tqdm
except ImportError:
    def tqdm(x, **kwargs):
        return x

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from torchvision.models import resnet18

from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.decomposition import PCA

# ============================================================
# Paths
# ============================================================
PROJECT_ROOT = os.getcwd()
DATA_ROOT = os.path.join(PROJECT_ROOT, "shift_aware_conformal_data")
OUTPUT_ROOT = os.path.join(PROJECT_ROOT, "shift_aware_conformal_outputs")
FIG_DIR = os.path.join(OUTPUT_ROOT, "step4_weighted_conformal_figures_stabilized")

os.makedirs(FIG_DIR, exist_ok=True)

CAL_PATH = os.path.join(OUTPUT_ROOT, "cal_outputs_full.npz")
MODEL_PATH = os.path.join(OUTPUT_ROOT, "resnet18_cifar10_best_debug.pt")
CIFAR10C_ROOT = os.path.join(DATA_ROOT, "CIFAR-10-C")

RESULTS_CSV = os.path.join(OUTPUT_ROOT, "step4_weighted_conformal_results_stabilized.csv")
RESULTS_NPZ = os.path.join(OUTPUT_ROOT, "step4_weighted_conformal_results_stabilized.npz")

print("PROJECT_ROOT:", PROJECT_ROOT)
print("DATA_ROOT   :", DATA_ROOT)
print("OUTPUT_ROOT :", OUTPUT_ROOT)
print("FIG_DIR     :", FIG_DIR)
print("CAL_PATH    :", CAL_PATH)
print("MODEL_PATH  :", MODEL_PATH)
print("CIFAR10C    :", CIFAR10C_ROOT)

# ============================================================
# Checks
# ============================================================
for path in [CAL_PATH, MODEL_PATH]:
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing required file: {path}")

if not os.path.exists(CIFAR10C_ROOT):
    raise FileNotFoundError(f"Missing CIFAR-10-C folder: {CIFAR10C_ROOT}")

# ============================================================
# Device
# ============================================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device      :", device)

# ============================================================
# Settings
# ============================================================
BATCH_SIZE = 128
NUM_WORKERS = 0
NUM_CLASSES = 10
SEED = 42

ALPHAS = [0.10, 0.05]

CORRUPTIONS = [
    "gaussian_noise",
    "motion_blur",
    "brightness",
    "jpeg_compression",
]
SEVERITIES = [1, 2, 3, 4, 5]

# RAPS settings
METHOD_NAME = "WCP-RAPS-stabilized"
LAMBDA_REG = 0.01
K_REG = 2
RANDOMIZED = True

# Stabilized weighting / safety settings
N_SPLITS = 5
W_CLIP = 5.0
ESS_RATIO_THRESHOLD = 0.30
MAX_WEIGHT_AFTER_NORM = 15.0
LOGREG_C = 0.1
PCA_COMPONENTS = 50

# ============================================================
# Reproducibility
# ============================================================
np.random.seed(SEED)
torch.manual_seed(SEED)

# ============================================================
# Model
# ============================================================
class ResNet18CIFAR(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.backbone = resnet18(weights=None)
        self.backbone.conv1 = nn.Conv2d(
            3, 64, kernel_size=3, stride=1, padding=1, bias=False
        )
        self.backbone.maxpool = nn.Identity()
        in_features = self.backbone.fc.in_features
        self.backbone.fc = nn.Linear(in_features, num_classes)

    def forward(self, x):
        return self.backbone(x)

    def forward_with_features(self, x):
        x = self.backbone.conv1(x)
        x = self.backbone.bn1(x)
        x = self.backbone.relu(x)
        x = self.backbone.maxpool(x)

        x = self.backbone.layer1(x)
        x = self.backbone.layer2(x)
        x = self.backbone.layer3(x)
        x = self.backbone.layer4(x)

        x = self.backbone.avgpool(x)
        features = torch.flatten(x, 1)
        logits = self.backbone.fc(features)
        return logits, features

# ============================================================
# CIFAR-10-C dataset
# ============================================================
class CIFAR10CDataset(Dataset):
    def __init__(self, cifar10c_root, corruption_name, severity, transform=None):
        self.transform = transform

        corruption_path = os.path.join(cifar10c_root, f"{corruption_name}.npy")
        labels_path = os.path.join(cifar10c_root, "labels.npy")

        if not os.path.exists(corruption_path):
            raise FileNotFoundError(f"Missing corruption file: {corruption_path}")
        if not os.path.exists(labels_path):
            raise FileNotFoundError(f"Missing labels file: {labels_path}")

        all_images = np.load(corruption_path)
        all_labels = np.load(labels_path)

        start = (severity - 1) * 10000
        end = severity * 10000

        self.images = all_images[start:end]
        self.labels = all_labels[start:end]

        assert len(self.images) == 10000
        assert len(self.labels) == 10000

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        img = self.images[idx]
        label = int(self.labels[idx])

        if self.transform is not None:
            img = self.transform(img)

        return img, label

# ============================================================
# Transform
# ============================================================
transform_cifar10c = transforms.Compose([
    transforms.ToPILImage(),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=(0.4914, 0.4822, 0.4465),
        std=(0.2470, 0.2435, 0.2616),
    ),
])

# ============================================================
# RAPS helpers
# ============================================================
def conformal_quantile(scores, alpha):
    n = len(scores)
    k = int(np.ceil((n + 1) * (1 - alpha))) - 1
    k = min(max(k, 0), n - 1)
    return np.sort(scores)[k]

def weighted_quantile(scores, weights, gamma):
    order = np.argsort(scores)
    s = scores[order]
    w = weights[order]

    cum_w = np.cumsum(w)
    total_w = np.sum(w)
    threshold = gamma * total_w

    idx = np.searchsorted(cum_w, threshold, side="left")
    idx = min(max(idx, 0), len(s) - 1)
    return float(s[idx])

def compute_raps_score(prob_vec, true_label, lambda_reg=0.0, k_reg=0, randomized=True, rng=None):
    if rng is None:
        rng = np.random.default_rng()

    sorted_idx = np.argsort(-prob_vec)
    sorted_probs = prob_vec[sorted_idx]

    L = int(np.where(sorted_idx == true_label)[0][0]) + 1

    score = np.sum(sorted_probs[:L]) + lambda_reg * max(L - k_reg, 0)

    if randomized:
        u = rng.uniform()
        score -= u * sorted_probs[L - 1]

    return float(score)

def compute_all_raps_scores(probs, labels, lambda_reg=0.0, k_reg=0, randomized=True, seed=0):
    rng = np.random.default_rng(seed)
    scores = []
    for i in range(len(labels)):
        scores.append(
            compute_raps_score(
                probs[i],
                labels[i],
                lambda_reg=lambda_reg,
                k_reg=k_reg,
                randomized=randomized,
                rng=rng,
            )
        )
    return np.array(scores, dtype=np.float64)

def build_raps_set(prob_vec, qhat, lambda_reg=0.0, k_reg=0, randomized=True, rng=None):
    if rng is None:
        rng = np.random.default_rng()

    sorted_idx = np.argsort(-prob_vec)
    sorted_probs = prob_vec[sorted_idx]
    K = len(sorted_probs)

    penalized_cum = np.array([
        np.sum(sorted_probs[:j]) + lambda_reg * max(j - k_reg, 0)
        for j in range(1, K + 1)
    ], dtype=np.float64)

    count_leq = int(np.sum(penalized_cum <= qhat))

    if count_leq >= K:
        return [int(x) for x in sorted_idx]

    L = count_leq + 1

    if randomized:
        boundary_value = penalized_cum[L - 1]
        increment = sorted_probs[L - 1] + (lambda_reg if L > k_reg else 0.0)

        drop_prob = (boundary_value - qhat) / max(increment, 1e-12)
        drop_prob = float(np.clip(drop_prob, 0.0, 1.0))

        u = rng.uniform()
        if u <= drop_prob:
            L -= 1

    return [int(x) for x in sorted_idx[:L]]

def build_all_prediction_sets(probs, qhat, lambda_reg=0.0, k_reg=0, randomized=True, seed=0):
    rng = np.random.default_rng(seed)
    pred_sets = []
    for prob_vec in probs:
        pred_sets.append(
            build_raps_set(
                prob_vec,
                qhat=qhat,
                lambda_reg=lambda_reg,
                k_reg=k_reg,
                randomized=randomized,
                rng=rng,
            )
        )
    return pred_sets

def empirical_coverage(pred_sets, labels):
    return float(np.mean([int(y) in s for s, y in zip(pred_sets, labels)]))

def average_set_size(pred_sets):
    return float(np.mean([len(s) for s in pred_sets]))

# ============================================================
# Weight helpers
# ============================================================
def compute_ess(weights):
    return float((np.sum(weights) ** 2) / np.sum(weights ** 2))

def fit_oof_density_ratio_weights(
    cal_features,
    shift_features,
    n_splits=5,
    seed=42,
    w_clip=5.0,
    logreg_c=0.1,
    pca_components=50
):
    rng = np.random.default_rng(seed)
    oof_probs = np.zeros(len(cal_features), dtype=np.float64)

    kf = KFold(n_splits=n_splits, shuffle=True, random_state=seed)

    for fold, (train_idx, val_idx) in enumerate(kf.split(cal_features), start=1):
        X_cal_train = cal_features[train_idx]
        X_cal_val = cal_features[val_idx]

        m = len(train_idx)
        shift_idx = rng.choice(len(shift_features), size=min(m, len(shift_features)), replace=False)
        X_shift_train = shift_features[shift_idx]

        X_train = np.concatenate([X_cal_train, X_shift_train], axis=0)
        y_train = np.concatenate([
            np.zeros(len(X_cal_train), dtype=np.int64),
            np.ones(len(X_shift_train), dtype=np.int64)
        ], axis=0)

        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_val_scaled = scaler.transform(X_cal_val)

        n_components = min(
            pca_components,
            X_train_scaled.shape[1],
            X_train_scaled.shape[0] - 1
        )
        pca = PCA(n_components=n_components, random_state=seed + fold)
        X_train_pca = pca.fit_transform(X_train_scaled)
        X_val_pca = pca.transform(X_val_scaled)

        clf = LogisticRegression(
            penalty="l2",
            C=logreg_c,
            solver="lbfgs",
            max_iter=1000,
            class_weight="balanced",
            random_state=seed + fold,
        )
        clf.fit(X_train_pca, y_train)

        p_val = clf.predict_proba(X_val_pca)[:, 1]
        p_val = np.clip(p_val, 1e-4, 1 - 1e-4)
        oof_probs[val_idx] = p_val

    raw_weights = oof_probs / (1.0 - oof_probs)
    raw_weights = np.minimum(raw_weights, w_clip)
    norm_weights = raw_weights / np.mean(raw_weights)

    ess = compute_ess(norm_weights)

    diagnostics = {
        "domain_prob_mean": float(np.mean(oof_probs)),
        "domain_prob_std": float(np.std(oof_probs)),
        "raw_weight_mean": float(np.mean(raw_weights)),
        "raw_weight_std": float(np.std(raw_weights)),
        "norm_weight_mean": float(np.mean(norm_weights)),
        "norm_weight_std": float(np.std(norm_weights)),
        "min_weight": float(np.min(norm_weights)),
        "max_weight": float(np.max(norm_weights)),
        "ess": ess,
        "ess_ratio": ess / len(norm_weights),
    }

    return norm_weights.astype(np.float64), diagnostics

# ============================================================
# Inference helper
# ============================================================
def collect_probs_features_labels(model, loader, device, desc="Collect"):
    model.eval()
    all_probs = []
    all_features = []
    all_labels = []

    with torch.no_grad():
        for images, labels in tqdm(loader, desc=desc, leave=False):
            images = images.to(device)
            logits, features = model.forward_with_features(images)
            probs = torch.softmax(logits, dim=1)

            all_probs.append(probs.cpu().numpy())
            all_features.append(features.cpu().numpy())
            all_labels.append(labels.numpy())

    probs = np.concatenate(all_probs, axis=0)
    features = np.concatenate(all_features, axis=0)
    labels = np.concatenate(all_labels, axis=0)
    return probs, features, labels

# ============================================================
# Load clean calibration data
# ============================================================
cal_data = np.load(CAL_PATH)

required_keys = ["probs", "labels", "features"]
for k in required_keys:
    if k not in cal_data.files:
        raise KeyError(f"'{k}' not found in {CAL_PATH}")

cal_probs = cal_data["probs"]
cal_labels = cal_data["labels"]
cal_features = cal_data["features"]

print("\nLoaded calibration arrays:")
print("cal_probs    :", cal_probs.shape)
print("cal_labels   :", cal_labels.shape)
print("cal_features :", cal_features.shape)

# ============================================================
# Compute clean calibration scores and standard qhat
# ============================================================
cal_scores = compute_all_raps_scores(
    cal_probs,
    cal_labels,
    lambda_reg=LAMBDA_REG,
    k_reg=K_REG,
    randomized=RANDOMIZED,
    seed=SEED,
)

qhat_standard = {}
for alpha in ALPHAS:
    qhat_standard[alpha] = conformal_quantile(cal_scores, alpha)
    print(f"alpha={alpha:.2f} | standard qhat={qhat_standard[alpha]:.6f}")

# ============================================================
# Load model
# ============================================================
model = ResNet18CIFAR(num_classes=NUM_CLASSES).to(device)
model.load_state_dict(torch.load(MODEL_PATH, map_location=device))
model.eval()
print("\nLoaded model successfully.")

# ============================================================
# Main loop
# ============================================================
all_results = []

for corruption in CORRUPTIONS:
    print("\n" + "=" * 90)
    print("Corruption:", corruption)

    for severity in SEVERITIES:
        dataset = CIFAR10CDataset(
            cifar10c_root=CIFAR10C_ROOT,
            corruption_name=corruption,
            severity=severity,
            transform=transform_cifar10c,
        )

        loader = DataLoader(
            dataset,
            batch_size=BATCH_SIZE,
            shuffle=False,
            num_workers=NUM_WORKERS,
        )

        shift_probs, shift_features, shift_labels = collect_probs_features_labels(
            model,
            loader,
            device,
            desc=f"{corruption} sev{severity}"
        )

        weights, diag = fit_oof_density_ratio_weights(
            cal_features=cal_features,
            shift_features=shift_features,
            n_splits=N_SPLITS,
            seed=SEED + severity,
            w_clip=W_CLIP,
            logreg_c=LOGREG_C,
            pca_components=PCA_COMPONENTS,
        )

        ess = diag["ess"]
        ess_ratio = diag["ess_ratio"]
        max_weight = diag["max_weight"]

        safe_to_weight = (ess_ratio >= ESS_RATIO_THRESHOLD) and (max_weight <= MAX_WEIGHT_AFTER_NORM)

        if safe_to_weight:
            fallback_reason = "weighted"
        elif ess_ratio < ESS_RATIO_THRESHOLD and max_weight > MAX_WEIGHT_AFTER_NORM:
            fallback_reason = "ess_and_max_weight"
        elif ess_ratio < ESS_RATIO_THRESHOLD:
            fallback_reason = "ess"
        else:
            fallback_reason = "max_weight"

        print(
            f"severity={severity} | "
            f"ess={ess:.2f} | ess_ratio={ess_ratio:.4f} | "
            f"min_weight={diag['min_weight']:.4f} | max_weight={max_weight:.4f} | "
            f"mode={fallback_reason}"
        )

        for alpha in ALPHAS:
            q_std = qhat_standard[alpha]
            q_w = weighted_quantile(cal_scores, weights, 1 - alpha)
            q_final = q_w if safe_to_weight else q_std

            pred_std = build_all_prediction_sets(
                shift_probs,
                qhat=q_std,
                lambda_reg=LAMBDA_REG,
                k_reg=K_REG,
                randomized=RANDOMIZED,
                seed=SEED + 100 + severity,
            )
            pred_w = build_all_prediction_sets(
                shift_probs,
                qhat=q_w,
                lambda_reg=LAMBDA_REG,
                k_reg=K_REG,
                randomized=RANDOMIZED,
                seed=SEED + 200 + severity,
            )
            pred_final = build_all_prediction_sets(
                shift_probs,
                qhat=q_final,
                lambda_reg=LAMBDA_REG,
                k_reg=K_REG,
                randomized=RANDOMIZED,
                seed=SEED + 300 + severity,
            )

            cov_std = empirical_coverage(pred_std, shift_labels)
            size_std = average_set_size(pred_std)

            cov_w = empirical_coverage(pred_w, shift_labels)
            size_w = average_set_size(pred_w)

            cov_final = empirical_coverage(pred_final, shift_labels)
            size_final = average_set_size(pred_final)

            row = {
                "method": METHOD_NAME,
                "corruption": corruption,
                "severity": severity,
                "alpha": alpha,
                "target_coverage": 1 - alpha,

                "qhat_standard": q_std,
                "qhat_weighted": q_w,
                "qhat_final": q_final,

                "coverage_standard": cov_std,
                "coverage_gap_standard": cov_std - (1 - alpha),
                "avg_set_size_standard": size_std,

                "coverage_weighted": cov_w,
                "coverage_gap_weighted": cov_w - (1 - alpha),
                "avg_set_size_weighted": size_w,

                "coverage_final": cov_final,
                "coverage_gap_final": cov_final - (1 - alpha),
                "avg_set_size_final": size_final,

                "safe_to_weight": int(safe_to_weight),
                "fallback_reason": fallback_reason,

                "ess": ess,
                "ess_ratio": ess_ratio,
                "min_weight": diag["min_weight"],
                "max_weight": max_weight,
                "mean_weight": float(np.mean(weights)),
                "std_weight": float(np.std(weights)),
                "domain_prob_mean": diag["domain_prob_mean"],
                "domain_prob_std": diag["domain_prob_std"],
            }
            all_results.append(row)

            print(
                f"  alpha={alpha:.2f} | "
                f"std_cov={cov_std:.4f}, w_cov={cov_w:.4f}, final_cov={cov_final:.4f} | "
                f"std_size={size_std:.4f}, w_size={size_w:.4f}, final_size={size_final:.4f}"
            )

# ============================================================
# Save CSV
# ============================================================
fieldnames = list(all_results[0].keys())

with open(RESULTS_CSV, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(all_results)

print("\nSaved CSV:", RESULTS_CSV)

# ============================================================
# Save NPZ
# ============================================================
np.savez_compressed(
    RESULTS_NPZ,
    results=np.array(all_results, dtype=object)
)

print("Saved NPZ:", RESULTS_NPZ)

# ============================================================
# Summary print
# ============================================================
print("\n" + "=" * 90)
print("STEP 4 SUMMARY")
print("=" * 90)

for alpha in ALPHAS:
    print(f"\nAlpha = {alpha:.2f}")
    print("-" * 80)

    for corruption in CORRUPTIONS:
        rows = [
            r for r in all_results
            if r["corruption"] == corruption and abs(r["alpha"] - alpha) < 1e-12
        ]

        rows = sorted(rows, key=lambda x: x["severity"])

        best_final_cov = max(r["coverage_final"] for r in rows)
        worst_final_cov = min(r["coverage_final"] for r in rows)
        best_ess_ratio = max(r["ess_ratio"] for r in rows)
        min_max_weight = min(r["max_weight"] for r in rows)
        max_max_weight = max(r["max_weight"] for r in rows)
        used_weight_count = sum(r["safe_to_weight"] for r in rows)

        print(
            f"{corruption:20s} | worst final cov = {worst_final_cov:.4f} | "
            f"best final cov = {best_final_cov:.4f} | "
            f"best ESS ratio = {best_ess_ratio:.4f} | "
            f"max_weight range = [{min_max_weight:.2f}, {max_max_weight:.2f}] | "
            f"weighted used in {used_weight_count}/5 severities"
        )

# ============================================================
# Figures
# ============================================================
for alpha in ALPHAS:
    for corruption in CORRUPTIONS:
        rows = [
            r for r in all_results
            if r["corruption"] == corruption and abs(r["alpha"] - alpha) < 1e-12
        ]
        rows = sorted(rows, key=lambda x: x["severity"])

        xs = [r["severity"] for r in rows]
        ys_std = [r["coverage_standard"] for r in rows]
        ys_w = [r["coverage_weighted"] for r in rows]
        ys_final = [r["coverage_final"] for r in rows]

        plt.figure(figsize=(8, 5))
        plt.plot(xs, ys_std, marker="o", label="Standard RAPS")
        plt.plot(xs, ys_w, marker="o", label="Weighted RAPS")
        plt.plot(xs, ys_final, marker="o", label="Safe final")
        plt.axhline(1 - alpha, linestyle="--", label=f"target={1-alpha:.2f}")
        plt.xlabel("Severity")
        plt.ylabel("Coverage")
        plt.title(f"{corruption} | coverage vs severity | alpha={alpha:.2f}")
        plt.xticks(SEVERITIES)
        plt.ylim(0, 1.05)
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.savefig(
            os.path.join(FIG_DIR, f"coverage_{corruption}_alpha_{str(alpha).replace('.', 'p')}.png"),
            dpi=300,
            bbox_inches="tight",
        )
        plt.show()

    for corruption in CORRUPTIONS:
        rows = [
            r for r in all_results
            if r["corruption"] == corruption and abs(r["alpha"] - alpha) < 1e-12
        ]
        rows = sorted(rows, key=lambda x: x["severity"])

        xs = [r["severity"] for r in rows]
        ys_std = [r["avg_set_size_standard"] for r in rows]
        ys_w = [r["avg_set_size_weighted"] for r in rows]
        ys_final = [r["avg_set_size_final"] for r in rows]

        plt.figure(figsize=(8, 5))
        plt.plot(xs, ys_std, marker="o", label="Standard RAPS")
        plt.plot(xs, ys_w, marker="o", label="Weighted RAPS")
        plt.plot(xs, ys_final, marker="o", label="Safe final")
        plt.xlabel("Severity")
        plt.ylabel("Average set size")
        plt.title(f"{corruption} | set size vs severity | alpha={alpha:.2f}")
        plt.xticks(SEVERITIES)
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.savefig(
            os.path.join(FIG_DIR, f"setsize_{corruption}_alpha_{str(alpha).replace('.', 'p')}.png"),
            dpi=300,
            bbox_inches="tight",
        )
        plt.show()

for corruption in CORRUPTIONS:
    rows = [
        r for r in all_results
        if r["corruption"] == corruption and abs(r["alpha"] - 0.10) < 1e-12
    ]
    rows = sorted(rows, key=lambda x: x["severity"])

    xs = [r["severity"] for r in rows]
    ys = [r["ess_ratio"] for r in rows]
    modes = [r["safe_to_weight"] for r in rows]

    plt.figure(figsize=(8, 5))
    plt.plot(xs, ys, marker="o")
    plt.axhline(ESS_RATIO_THRESHOLD, linestyle="--", label=f"ESS threshold={ESS_RATIO_THRESHOLD:.2f}")
    for x, y, m in zip(xs, ys, modes):
        plt.text(x, y, "W" if m == 1 else "F", fontsize=10, ha="center", va="bottom")
    plt.xlabel("Severity")
    plt.ylabel("ESS / n")
    plt.title(f"{corruption} | ESS ratio vs severity")
    plt.xticks(SEVERITIES)
    plt.ylim(0, 1.05)
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(
        os.path.join(FIG_DIR, f"ess_ratio_{corruption}.png"),
        dpi=300,
        bbox_inches="tight",
    )
    plt.show()

for corruption in CORRUPTIONS:
    rows = [
        r for r in all_results
        if r["corruption"] == corruption and abs(r["alpha"] - 0.10) < 1e-12
    ]
    rows = sorted(rows, key=lambda x: x["severity"])

    xs = [r["severity"] for r in rows]
    ys = [r["max_weight"] for r in rows]
    modes = [r["safe_to_weight"] for r in rows]

    plt.figure(figsize=(8, 5))
    plt.plot(xs, ys, marker="o")
    plt.axhline(MAX_WEIGHT_AFTER_NORM, linestyle="--", label=f"max-weight threshold={MAX_WEIGHT_AFTER_NORM:.2f}")
    for x, y, m in zip(xs, ys, modes):
        plt.text(x, y, "W" if m == 1 else "F", fontsize=10, ha="center", va="bottom")
    plt.xlabel("Severity")
    plt.ylabel("Max normalized weight")
    plt.title(f"{corruption} | max weight vs severity")
    plt.xticks(SEVERITIES)
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(
        os.path.join(FIG_DIR, f"max_weight_{corruption}.png"),
        dpi=300,
        bbox_inches="tight",
    )
    plt.show()

print("\nSaved figures in:", FIG_DIR)
print("Step 4 finished.")
```
