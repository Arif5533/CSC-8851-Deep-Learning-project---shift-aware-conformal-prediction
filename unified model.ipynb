```python
import os
import csv
import numpy as np
import matplotlib.pyplot as plt

try:
    from tqdm import tqdm
except ImportError:
    def tqdm(x, **kwargs):
        return x

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from torchvision.models import resnet18

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import KFold
from sklearn.decomposition import PCA
from sklearn.covariance import EmpiricalCovariance


PROJECT_ROOT = os.getcwd()
DATA_ROOT = os.path.join(PROJECT_ROOT, "shift_aware_conformal_data")
OUTPUT_ROOT = os.path.join(PROJECT_ROOT, "shift_aware_conformal_outputs")
FIG_DIR = os.path.join(OUTPUT_ROOT, "step6_unified_system_figures")

os.makedirs(FIG_DIR, exist_ok=True)

CAL_PATH = os.path.join(OUTPUT_ROOT, "cal_outputs_full.npz")
MODEL_PATH = os.path.join(OUTPUT_ROOT, "resnet18_cifar10_best_debug.pt")
CIFAR10C_ROOT = os.path.join(DATA_ROOT, "CIFAR-10-C")

# Optional EDL files
EDL_CAL_PATH = os.path.join(OUTPUT_ROOT, "edl_cal_outputs.npz")

RESULTS_CSV = os.path.join(OUTPUT_ROOT, "step6_unified_system_results.csv")
SAMPLE_CSV = os.path.join(OUTPUT_ROOT, "step6_unified_system_sample_outputs.csv")
RESULTS_NPZ = os.path.join(OUTPUT_ROOT, "step6_unified_system_results.npz")

print("PROJECT_ROOT:", PROJECT_ROOT)
print("DATA_ROOT   :", DATA_ROOT)
print("OUTPUT_ROOT :", OUTPUT_ROOT)
print("CAL_PATH    :", CAL_PATH)
print("MODEL_PATH  :", MODEL_PATH)
print("CIFAR10C    :", CIFAR10C_ROOT)
print("EDL_CAL     :", EDL_CAL_PATH)


for path in [CAL_PATH, MODEL_PATH]:
    if not os.path.exists(path):
        raise FileNotFoundError(f"Missing required file: {path}")

if not os.path.exists(CIFAR10C_ROOT):
    raise FileNotFoundError(f"Missing CIFAR-10-C folder: {CIFAR10C_ROOT}")


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device      :", device)

SEED = 42
BATCH_SIZE = 128
NUM_WORKERS = 0
NUM_CLASSES = 10

# Main conformal / gate settings
ALPHA = 0.10
DELTA = 0.10
LAMBDA_REG = 0.01
K_REG = 2
RANDOMIZED = True

# Stabilized weighting settings (match Step 4 stabilized version)
N_SPLITS = 5
W_CLIP = 5.0
ESS_RATIO_THRESHOLD = 0.30
MAX_WEIGHT_AFTER_NORM = 15.0
LOGREG_C = 0.1
PCA_COMPONENTS = 50

# Main gate mode for deployed system
# choices: "energy", "mahalanobis"
GATE_MODE = "energy"

CORRUPTIONS = [
    "gaussian_noise",
    "motion_blur",
    "brightness",
    "jpeg_compression",
]
SEVERITIES = [1, 2, 3, 4, 5]

np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


class ResNet18CIFAR(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.backbone = resnet18(weights=None)
        self.backbone.conv1 = nn.Conv2d(
            3, 64, kernel_size=3, stride=1, padding=1, bias=False
        )
        self.backbone.maxpool = nn.Identity()
        in_features = self.backbone.fc.in_features
        self.backbone.fc = nn.Linear(in_features, num_classes)

    def forward(self, x):
        return self.backbone(x)

    def forward_with_features(self, x):
        x = self.backbone.conv1(x)
        x = self.backbone.bn1(x)
        x = self.backbone.relu(x)
        x = self.backbone.maxpool(x)

        x = self.backbone.layer1(x)
        x = self.backbone.layer2(x)
        x = self.backbone.layer3(x)
        x = self.backbone.layer4(x)

        x = self.backbone.avgpool(x)
        features = torch.flatten(x, 1)
        logits = self.backbone.fc(features)
        return logits, features


# CIFAR-10-C dataset

class CIFAR10CDataset(Dataset):
    def __init__(self, cifar10c_root, corruption_name, severity, transform=None):
        self.transform = transform

        corruption_path = os.path.join(cifar10c_root, f"{corruption_name}.npy")
        labels_path = os.path.join(cifar10c_root, "labels.npy")

        if not os.path.exists(corruption_path):
            raise FileNotFoundError(f"Missing corruption file: {corruption_path}")
        if not os.path.exists(labels_path):
            raise FileNotFoundError(f"Missing labels file: {labels_path}")

        all_images = np.load(corruption_path)
        all_labels = np.load(labels_path)

        start = (severity - 1) * 10000
        end = severity * 10000

        self.images = all_images[start:end]
        self.labels = all_labels[start:end]

        assert len(self.images) == 10000
        assert len(self.labels) == 10000

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        img = self.images[idx]
        label = int(self.labels[idx])

        if self.transform is not None:
            img = self.transform(img)

        return img, label


# Transform

transform_cifar10c = transforms.Compose([
    transforms.ToPILImage(),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=(0.4914, 0.4822, 0.4465),
        std=(0.2470, 0.2435, 0.2616),
    ),
])


# RAPS helpers

def conformal_quantile(scores, alpha):
    n = len(scores)
    k = int(np.ceil((n + 1) * (1 - alpha))) - 1
    k = min(max(k, 0), n - 1)
    return float(np.sort(scores)[k])

def weighted_quantile(scores, weights, gamma):
    order = np.argsort(scores)
    s = scores[order]
    w = weights[order]

    cum_w = np.cumsum(w)
    total_w = np.sum(w)
    threshold = gamma * total_w

    idx = np.searchsorted(cum_w, threshold, side="left")
    idx = min(max(idx, 0), len(s) - 1)
    return float(s[idx])

def compute_raps_score(prob_vec, true_label, lambda_reg=0.0, k_reg=0, randomized=True, rng=None):
    if rng is None:
        rng = np.random.default_rng()

    sorted_idx = np.argsort(-prob_vec)
    sorted_probs = prob_vec[sorted_idx]
    L = int(np.where(sorted_idx == true_label)[0][0]) + 1

    score = np.sum(sorted_probs[:L]) + lambda_reg * max(L - k_reg, 0)

    if randomized:
        u = rng.uniform()
        score -= u * sorted_probs[L - 1]

    return float(score)

def compute_all_raps_scores(probs, labels, lambda_reg=0.0, k_reg=0, randomized=True, seed=0):
    rng = np.random.default_rng(seed)
    scores = []
    for i in range(len(labels)):
        scores.append(
            compute_raps_score(
                probs[i],
                labels[i],
                lambda_reg=lambda_reg,
                k_reg=k_reg,
                randomized=randomized,
                rng=rng,
            )
        )
    return np.array(scores, dtype=np.float64)

def build_raps_set(prob_vec, qhat, lambda_reg=0.0, k_reg=0, randomized=True, rng=None):
    if rng is None:
        rng = np.random.default_rng()

    sorted_idx = np.argsort(-prob_vec)
    sorted_probs = prob_vec[sorted_idx]
    K = len(sorted_probs)

    penalized_cum = np.array([
        np.sum(sorted_probs[:j]) + lambda_reg * max(j - k_reg, 0)
        for j in range(1, K + 1)
    ], dtype=np.float64)

    count_leq = int(np.sum(penalized_cum <= qhat))

    if count_leq >= K:
        return [int(x) for x in sorted_idx]

    L = count_leq + 1

    if randomized:
        boundary_value = penalized_cum[L - 1]
        increment = sorted_probs[L - 1] + (lambda_reg if L > k_reg else 0.0)
        drop_prob = (boundary_value - qhat) / max(increment, 1e-12)
        drop_prob = float(np.clip(drop_prob, 0.0, 1.0))

        u = rng.uniform()
        if u <= drop_prob:
            L -= 1

    return [int(x) for x in sorted_idx[:L]]

def build_all_prediction_sets(probs, qhat, lambda_reg=0.0, k_reg=0, randomized=True, seed=0):
    rng = np.random.default_rng(seed)
    return [
        build_raps_set(
            prob_vec,
            qhat=qhat,
            lambda_reg=lambda_reg,
            k_reg=k_reg,
            randomized=randomized,
            rng=rng,
        )
        for prob_vec in probs
    ]


# Score helpers

def energy_score_from_logits(logits, T=1.0):
    logits_t = torch.tensor(logits, dtype=torch.float32) / T
    energy = -T * torch.logsumexp(logits_t, dim=1)
    return energy.numpy()

def conformal_upper_quantile(scores, delta):
    n = len(scores)
    k = int(np.ceil((n + 1) * (1 - delta))) - 1
    k = min(max(k, 0), n - 1)
    return float(np.sort(scores)[k])

def compute_ess(weights):
    return float((np.sum(weights) ** 2) / np.sum(weights ** 2))


# Stabilized weight estimator

def fit_oof_density_ratio_weights(
    cal_features,
    shift_features,
    n_splits=5,
    seed=42,
    w_clip=5.0,
    logreg_c=0.1,
    pca_components=50
):
    rng = np.random.default_rng(seed)
    oof_probs = np.zeros(len(cal_features), dtype=np.float64)

    kf = KFold(n_splits=n_splits, shuffle=True, random_state=seed)

    for fold, (train_idx, val_idx) in enumerate(kf.split(cal_features), start=1):
        X_cal_train = cal_features[train_idx]
        X_cal_val = cal_features[val_idx]

        m = len(train_idx)
        shift_idx = rng.choice(len(shift_features), size=min(m, len(shift_features)), replace=False)
        X_shift_train = shift_features[shift_idx]

        X_train = np.concatenate([X_cal_train, X_shift_train], axis=0)
        y_train = np.concatenate([
            np.zeros(len(X_cal_train), dtype=np.int64),
            np.ones(len(X_shift_train), dtype=np.int64)
        ], axis=0)

        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_val_scaled = scaler.transform(X_cal_val)

        n_components = min(
            pca_components,
            X_train_scaled.shape[1],
            X_train_scaled.shape[0] - 1
        )
        pca = PCA(n_components=n_components, random_state=seed + fold)
        X_train_pca = pca.fit_transform(X_train_scaled)
        X_val_pca = pca.transform(X_val_scaled)

        clf = LogisticRegression(
            penalty="l2",
            C=logreg_c,
            solver="lbfgs",
            max_iter=1000,
            class_weight="balanced",
            random_state=seed + fold,
        )
        clf.fit(X_train_pca, y_train)

        p_val = clf.predict_proba(X_val_pca)[:, 1]
        p_val = np.clip(p_val, 1e-4, 1 - 1e-4)
        oof_probs[val_idx] = p_val

    raw_weights = oof_probs / (1.0 - oof_probs)
    raw_weights = np.minimum(raw_weights, w_clip)
    norm_weights = raw_weights / np.mean(raw_weights)

    ess = compute_ess(norm_weights)
    diagnostics = {
        "domain_prob_mean": float(np.mean(oof_probs)),
        "domain_prob_std": float(np.std(oof_probs)),
        "max_weight": float(np.max(norm_weights)),
        "ess": ess,
        "ess_ratio": ess / len(norm_weights),
    }
    return norm_weights.astype(np.float64), diagnostics


# Mahalanobis helper

def mahalanobis_score(features, class_means, precision):
    dists = []
    for c in range(class_means.shape[0]):
        diff = features - class_means[c]
        dist_c = np.sum((diff @ precision) * diff, axis=1)
        dists.append(dist_c)
    dists = np.stack(dists, axis=1)
    return np.min(dists, axis=1)


# Inference helper

def collect_probs_logits_features_labels(model, loader, device, desc="Collect"):
    model.eval()
    all_probs = []
    all_logits = []
    all_features = []
    all_labels = []

    with torch.no_grad():
        for images, labels in tqdm(loader, desc=desc, leave=False):
            images = images.to(device)
            logits, features = model.forward_with_features(images)
            probs = torch.softmax(logits, dim=1)

            all_probs.append(probs.cpu().numpy())
            all_logits.append(logits.cpu().numpy())
            all_features.append(features.cpu().numpy())
            all_labels.append(labels.numpy())

    probs = np.concatenate(all_probs, axis=0)
    logits = np.concatenate(all_logits, axis=0)
    features = np.concatenate(all_features, axis=0)
    labels = np.concatenate(all_labels, axis=0)
    return probs, logits, features, labels


# Load calibration outputs

cal_data = np.load(CAL_PATH)

required_keys = ["probs", "labels", "features", "logits"]
for k in required_keys:
    if k not in cal_data.files:
        raise KeyError(f"'{k}' not found in {CAL_PATH}")

cal_probs = cal_data["probs"]
cal_labels = cal_data["labels"]
cal_features = cal_data["features"]
cal_logits = cal_data["logits"]

print("\nLoaded calibration arrays:")
print("cal_probs    :", cal_probs.shape)
print("cal_labels   :", cal_labels.shape)
print("cal_features :", cal_features.shape)
print("cal_logits   :", cal_logits.shape)


# Standard RAPS calibration

cal_raps_scores = compute_all_raps_scores(
    cal_probs,
    cal_labels,
    lambda_reg=LAMBDA_REG,
    k_reg=K_REG,
    randomized=RANDOMIZED,
    seed=SEED,
)
qhat_standard = conformal_quantile(cal_raps_scores, ALPHA)
print(f"\nStandard RAPS qhat (alpha={ALPHA:.2f}): {qhat_standard:.6f}")


# Energy gate threshold

energy_cal = energy_score_from_logits(cal_logits, T=1.0)
t_energy = conformal_upper_quantile(energy_cal, DELTA)
print(f"Energy gate threshold (delta={DELTA:.2f}): {t_energy:.6f}")


# Optional Mahalanobis gate threshold

class_means = None
precision = None
t_maha = None

try:
    class_means = []
    for c in range(NUM_CLASSES):
        cls_feats = cal_features[cal_labels == c]
        class_means.append(cls_feats.mean(axis=0))
    class_means = np.stack(class_means, axis=0)

    cov = EmpiricalCovariance(assume_centered=False)
    cov.fit(cal_features)
    precision = cov.precision_

    maha_cal = mahalanobis_score(cal_features, class_means, precision)
    t_maha = conformal_upper_quantile(maha_cal, DELTA)
    print(f"Mahalanobis gate threshold (delta={DELTA:.2f}): {t_maha:.6f}")
except Exception as e:
    print("Mahalanobis disabled:", e)


# Optional EDL threshold info

edl_available = os.path.exists(EDL_CAL_PATH)
if edl_available:
    edl_cal = np.load(EDL_CAL_PATH)
    edl_alpha0_cal = edl_cal["alpha0"]
    edl_gate_cal = -edl_alpha0_cal
    t_edl = conformal_upper_quantile(edl_gate_cal, DELTA)
    print(f"EDL gate threshold (delta={DELTA:.2f}): {t_edl:.6f}")
else:
    print("EDL calibration not found. EDL is treated as extension only.")


# Load model

model = ResNet18CIFAR(num_classes=NUM_CLASSES).to(device)
model.load_state_dict(torch.load(MODEL_PATH, map_location=device))
model.eval()
print("\nLoaded model successfully.")


# Unified prediction on one shift batch

def unified_predict_on_shift_batch(
    shift_probs,
    shift_logits,
    shift_features,
    shift_labels,
    corruption,
    severity,
    gate_mode="energy",
):
    # ----------------------------
    # Gate score
    # ----------------------------
    if gate_mode == "energy":
        gate_scores = energy_score_from_logits(shift_logits, T=1.0)
        gate_threshold = t_energy
    elif gate_mode == "mahalanobis":
        if class_means is None or precision is None or t_maha is None:
            raise ValueError("Mahalanobis gate requested but not available.")
        gate_scores = mahalanobis_score(shift_features, class_means, precision)
        gate_threshold = t_maha
    else:
        raise ValueError(f"Unsupported gate_mode: {gate_mode}")

    # ----------------------------
    # OOD gate
    # ----------------------------
    ood_flag = gate_scores >= gate_threshold

    # ----------------------------
    # Shift weights (batch-level)
    # ----------------------------
    weights, diag = fit_oof_density_ratio_weights(
        cal_features=cal_features,
        shift_features=shift_features,
        n_splits=N_SPLITS,
        seed=SEED + severity,
        w_clip=W_CLIP,
        logreg_c=LOGREG_C,
        pca_components=PCA_COMPONENTS,
    )

    ess = diag["ess"]
    ess_ratio = diag["ess_ratio"]
    max_weight = diag["max_weight"]

    weights_stable = (
        (ess_ratio >= ESS_RATIO_THRESHOLD) and
        (max_weight <= MAX_WEIGHT_AFTER_NORM)
    )

    qhat_weighted = weighted_quantile(cal_raps_scores, weights, 1 - ALPHA)

    # ----------------------------
    # Precompute both set types
    # ----------------------------
    pred_sets_standard = build_all_prediction_sets(
        shift_probs,
        qhat=qhat_standard,
        lambda_reg=LAMBDA_REG,
        k_reg=K_REG,
        randomized=RANDOMIZED,
        seed=SEED + 1000 + severity,
    )

    pred_sets_weighted = build_all_prediction_sets(
        shift_probs,
        qhat=qhat_weighted,
        lambda_reg=LAMBDA_REG,
        k_reg=K_REG,
        randomized=RANDOMIZED,
        seed=SEED + 2000 + severity,
    )

    # ----------------------------
    # Per-sample final decision
    # ----------------------------
    final_pred_sets = []
    final_statuses = []
    sample_rows = []

    for i in range(len(shift_labels)):
        y = int(shift_labels[i])

        if ood_flag[i]:
            status = "abstain/OOD"
            pred_set = []
        else:
            if weights_stable:
                status = "mild-shift(weighted)"
                pred_set = pred_sets_weighted[i]
            else:
                status = "fallback"
                pred_set = pred_sets_standard[i]

        final_pred_sets.append(pred_set)
        final_statuses.append(status)

        sample_rows.append({
            "corruption": corruption,
            "severity": severity,
            "sample_index": i,
            "true_label": y,
            "gate_score": float(gate_scores[i]),
            "gate_threshold": float(gate_threshold),
            "ood_flag": int(ood_flag[i]),
            "weights_stable": int(weights_stable),
            "status": status,
            "pred_set_size": len(pred_set),
            "pred_set": str(pred_set),
        })

    # ----------------------------
    # Summary
    # ----------------------------
    covered_non_abstain = 0
    total_non_abstain = 0
    total_abstain = 0

    for pred_set, y, status in zip(final_pred_sets, shift_labels, final_statuses):
        if status == "abstain/OOD":
            total_abstain += 1
        else:
            total_non_abstain += 1
            if int(y) in pred_set:
                covered_non_abstain += 1

    coverage_non_abstain = covered_non_abstain / max(total_non_abstain, 1)
    avg_set_size_non_abstain = (
        np.mean([len(s) for s, st in zip(final_pred_sets, final_statuses) if st != "abstain/OOD"])
        if total_non_abstain > 0 else 0.0
    )

    summary_row = {
        "corruption": corruption,
        "severity": severity,
        "alpha": ALPHA,
        "delta": DELTA,
        "gate_mode": gate_mode,

        "qhat_standard": qhat_standard,
        "qhat_weighted": qhat_weighted,

        "ess": ess,
        "ess_ratio": ess_ratio,
        "max_weight": max_weight,
        "weights_stable": int(weights_stable),

        "abstain_rate": total_abstain / len(shift_labels),
        "coverage_non_abstain": coverage_non_abstain,
        "coverage_gap_non_abstain": coverage_non_abstain - (1 - ALPHA),
        "avg_set_size_non_abstain": float(avg_set_size_non_abstain),

        "count_weighted": int(sum(st == "mild-shift(weighted)" for st in final_statuses)),
        "count_fallback": int(sum(st == "fallback" for st in final_statuses)),
        "count_abstain": int(sum(st == "abstain/OOD" for st in final_statuses)),
    }

    return summary_row, sample_rows


# Run unified system on CIFAR-10-C

all_summary_rows = []
all_sample_rows = []

for corruption in CORRUPTIONS:
    print("\n" + "=" * 90)
    print("Corruption:", corruption)

    for severity in SEVERITIES:
        dataset = CIFAR10CDataset(
            cifar10c_root=CIFAR10C_ROOT,
            corruption_name=corruption,
            severity=severity,
            transform=transform_cifar10c,
        )

        loader = DataLoader(
            dataset,
            batch_size=BATCH_SIZE,
            shuffle=False,
            num_workers=NUM_WORKERS,
        )

        shift_probs, shift_logits, shift_features, shift_labels = collect_probs_logits_features_labels(
            model,
            loader,
            device,
            desc=f"{corruption} sev{severity}"
        )

        summary_row, sample_rows = unified_predict_on_shift_batch(
            shift_probs=shift_probs,
            shift_logits=shift_logits,
            shift_features=shift_features,
            shift_labels=shift_labels,
            corruption=corruption,
            severity=severity,
            gate_mode=GATE_MODE,
        )

        all_summary_rows.append(summary_row)
        all_sample_rows.extend(sample_rows)

        print(
            f"severity={severity} | "
            f"abstain_rate={summary_row['abstain_rate']:.4f} | "
            f"cov_non_abstain={summary_row['coverage_non_abstain']:.4f} | "
            f"gap={summary_row['coverage_gap_non_abstain']:+.4f} | "
            f"avg_size={summary_row['avg_set_size_non_abstain']:.4f} | "
            f"ESS={summary_row['ess_ratio']:.4f} | "
            f"max_w={summary_row['max_weight']:.2f} | "
            f"stable={summary_row['weights_stable']}"
        )


with open(RESULTS_CSV, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=list(all_summary_rows[0].keys()))
    writer.writeheader()
    writer.writerows(all_summary_rows)

with open(SAMPLE_CSV, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=list(all_sample_rows[0].keys()))
    writer.writeheader()
    writer.writerows(all_sample_rows)

np.savez_compressed(
    RESULTS_NPZ,
    summary=np.array(all_summary_rows, dtype=object),
    samples=np.array(all_sample_rows, dtype=object),
)

print("\nSaved:")
print(" -", RESULTS_CSV)
print(" -", SAMPLE_CSV)
print(" -", RESULTS_NPZ)


print("\n" + "=" * 90)
print("FINAL UNIFIED SYSTEM SUMMARY")
print("=" * 90)

for corruption in CORRUPTIONS:
    rows = [r for r in all_summary_rows if r["corruption"] == corruption]
    rows = sorted(rows, key=lambda x: x["severity"])

    worst_gap = min(r["coverage_gap_non_abstain"] for r in rows)
    max_abstain = max(r["abstain_rate"] for r in rows)
    used_weighted = sum(r["weights_stable"] for r in rows)

    print(
        f"{corruption:20s} | "
        f"worst non-abstain gap = {worst_gap:+.4f} | "
        f"max abstain rate = {max_abstain:.4f} | "
        f"weighted stable in {used_weighted}/5 severities"
    )


# Abstain rate vs severity
for corruption in CORRUPTIONS:
    rows = [r for r in all_summary_rows if r["corruption"] == corruption]
    rows = sorted(rows, key=lambda x: x["severity"])

    xs = [r["severity"] for r in rows]
    ys = [r["abstain_rate"] for r in rows]

    plt.figure(figsize=(7, 4.5))
    plt.plot(xs, ys, marker="o")
    plt.xlabel("Severity")
    plt.ylabel("Abstain rate")
    plt.title(f"{corruption} | abstain rate vs severity")
    plt.xticks(SEVERITIES)
    plt.ylim(0, 1.0)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, f"abstain_rate_{corruption}.png"), dpi=300, bbox_inches="tight")
    plt.show()

# Coverage gap vs severity (non-abstain)
for corruption in CORRUPTIONS:
    rows = [r for r in all_summary_rows if r["corruption"] == corruption]
    rows = sorted(rows, key=lambda x: x["severity"])

    xs = [r["severity"] for r in rows]
    ys = [r["coverage_gap_non_abstain"] for r in rows]

    plt.figure(figsize=(7, 4.5))
    plt.plot(xs, ys, marker="o")
    plt.axhline(0.0, linestyle="--")
    plt.xlabel("Severity")
    plt.ylabel("Coverage gap (non-abstain)")
    plt.title(f"{corruption} | non-abstain coverage gap vs severity")
    plt.xticks(SEVERITIES)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, f"gap_non_abstain_{corruption}.png"), dpi=300, bbox_inches="tight")
    plt.show()

# Status counts stacked bar
for corruption in CORRUPTIONS:
    rows = [r for r in all_summary_rows if r["corruption"] == corruption]
    rows = sorted(rows, key=lambda x: x["severity"])

    xs = np.arange(len(rows))
    weighted_counts = np.array([r["count_weighted"] for r in rows])
    fallback_counts = np.array([r["count_fallback"] for r in rows])
    abstain_counts = np.array([r["count_abstain"] for r in rows])

    plt.figure(figsize=(8, 5))
    plt.bar(xs, weighted_counts, label="weighted")
    plt.bar(xs, fallback_counts, bottom=weighted_counts, label="fallback")
    plt.bar(xs, abstain_counts, bottom=weighted_counts + fallback_counts, label="abstain/OOD")
    plt.xticks(xs, [r["severity"] for r in rows])
    plt.xlabel("Severity")
    plt.ylabel("Count")
    plt.title(f"{corruption} | final status counts")
    plt.legend()
    plt.grid(True, axis="y", alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, f"status_counts_{corruption}.png"), dpi=300, bbox_inches="tight")
    plt.show()

print("\nSaved figures in:", FIG_DIR)
print("Step 6 unified system finished.")
```
